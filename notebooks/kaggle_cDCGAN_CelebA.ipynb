```python
# 3. Xác định đường dẫn tập dữ liệu CelebA trên Kaggle
import os

# Trên Kaggle, dataset CelebA thường nằm tại:
# /kaggle/input/celeba-dataset/img_align_celeba/img_align_celeba
# /kaggle/input/celeba-dataset/list_attr_celeba.csv hoặc list_attr_celeba.txt

KAGGLE_IMG_DIR = "/kaggle/input/celeba-dataset/img_align_celeba/img_align_celeba"
KAGGLE_ATTR_FILE = "/kaggle/input/celeba-dataset/list_attr_celeba.csv"

# Fallback tìm kiếm đường dẫn nếu khác cấu trúc:
if not os.path.exists(KAGGLE_IMG_DIR):
    for root, dirs, files in os.walk("/kaggle/input"):
        if "img_align_celeba" in dirs:
            KAGGLE_IMG_DIR = os.path.join(root, "img_align_celeba")
            break

print("Thư mục ảnh CelebA:", KAGGLE_IMG_DIR, "| Tồn tại:", os.path.exists(KAGGLE_IMG_DIR))
print("File thuộc tính:", KAGGLE_ATTR_FILE, "| Tồn tại:", os.path.exists(KAGGLE_ATTR_FILE))
```
